univariate plot

```python
import numpy as np
import matplotlib.pyplot as plt


flavours = ['Vanilla', 'Strawberry', 'Chocolate', 'MintChocolate', 'Others']
Frequency = [16,15,5,12,3]

c = ['r','y'] 

plt.bar(flavours,Frequency,width = 0.5,color = c,label = 'flavours')

plt.title('icecreme score')
plt.xlabel('FL')
plt.ylabel('FQ')
plt.legend()
plt.show()
```

bivariate

```python
age = [22,33,44]

weight = [28,67,89]



plt.scatter(age,weight,label='age',color = 'r')
plt.scatter(age,weight,label='weight',color = 'b')


plt.title('Age vs Weight')
plt.xlabel('Age')
plt.ylabel('Weight')
plt.legend()
plt.show()
```

multivariate

```python
x = ["math", "biology", "chemistry"]
a = [50, 60, 70]
b = [20, 20, 30]
c = [78, 45, 67]
d = [56, 34, 12]

width = 0.2

x_pos = np.arange(len(x)) 


plt.bar(x_pos, a, width=width, label='a')
plt.bar(x_pos + width, b, width=width, label='b')
plt.bar(x_pos + 2*width, c, width=width, label='c')
plt.bar(x_pos + 3*width, d, width=width, label='d')

plt.xticks(x_pos + 1.5*width, x)

plt.xlabel("Subjects")
plt.ylabel("Scores")
plt.title("Scores by Subject")
# plt.legend()

plt.show()
```
