#linear plot

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


days=[1,2,3,4,5,6,7]
temp= [24,25,26,27,28,29,30]

plt.plot(days,temp,color = 'red',marker = '*')

plt.title('temp checker')
plt.xlabel('Days')
plt.ylabel('Temperature')

plt.show()
```

```python
sub = ['math','urdu','eng','phy']

st1 = [30,44,56,78]
st2 = [77,89,90,45]

plt.plot(sub,st1,marker='*',color = 'r',label=st1)
plt.plot(sub,st2,marker ='o',color = 'b',label=st2)

plt.title('marks comparison')
plt.xlabel('sub')
plt.ylabel('marks')
plt.legend()
plt.show()
```

pi plot

```python
sub = ['math','urdu','bio','eng']
vote = [45,67,55,78]

plt.pie(vote,labels =sub,autopct = '%1.1f%%')

plt.title('favorte sub')
plt.show()
```

```python
fruits = ["Apple", "Banana", "Mango", "Orange", "Grapes"]
votes = [20, 25, 30, 10, 15]
explode = [0,0.1,0,0.2,0.3]

# (explode ka mtlb kitny prcnt plot s bahir o ga)

plt.pie(votes,labels=fruits,autopct="%1.1f%%",explode=explode)          

plt.show()
```
