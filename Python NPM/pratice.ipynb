```python
import numpy as np
import pandas as pd
```

```python
r = pd.read_csv('world_population.csv',usecols = ['2022 Population'])
max_val = r.max()
min_val = r.min()
print(max_val,min_val)
```

```python
wp = pd.read_csv('world_population.csv')

rowindex = wp['2022 Population'].idxmin()

country = wp['Country/Territory'][rowindex]
pop = wp['2022 Population'][rowindex]

print(country,pop)

# print(w)
```

find lowest rank of 10 countries

```python
r = pd.read_csv('world_population.csv')

low_rank = r.sort_values(by  ='Rank', ascending= False)

low_rank[['Rank','Country/Territory']].head(10)
```

```python
read_file = pd.read_csv('world_population.csv')

max_val = read_file['2022 Population'].idxmax()

country = read_file['Country/Territory'][max_val]

pop = read_file['2022 Population'][max_val]



print(country,pop,max_val)
```

find higest rank of 10 countries


```python
high_rank = read_file.sort_values(by = 'Rank',ascending = True)

high_rank[['Rank','Country/Territory','2020 Population']].head(20)
```
